```python
#quantitles banaty hyn pricve waly column ki
Q1 = data["price"].quantile(0.25)
Q3 = data["price"].quantile(0.75)
IQR = Q3 - Q1
```

```python
print(Q1, Q3, IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
#upper or lower limit ka formula likhty hy
```

```python
print(lower_limit, upper_limit)
```

```python
outliers = data[(data["price"] < lower_limit) | (data["price"] > upper_limit)]
outliers.shape[0]
# ab out liers ko nikalty hyn
```

```python
data["price"] = data["price"].clip(lower=lower_limit, upper=upper_limit)
# iskw stire krwa dety hyn us colmn me
```

```python
data["price"].max()
#iska max bh find krtyy hyn
```

```python
Q1 = data["minimum_nights"].quantile(0.25)
Q3 = data["minimum_nights"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[(data["minimum_nights"] < lower_limit) | (data["minimum_nights"] > upper_limit)]
print(outliers.shape[0])

data["minimum_nights"] = data["minimum_nights"].clip(lower=lower_limit, upper=upper_limit)
# same kam ab minumum nights k colmn k sath bh krty hyn
```

```python
Q1 = data["availability_365"].quantile(0.25)
Q3 = data["availability_365"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[(data["availability_365"] < lower_limit) | (data["availability_365"] > upper_limit)]
print(outliers.shape[0])

data["availability_365"] = data["availability_365"].clip(lower=lower_limit, upper=upper_limit)
#or same avaliabilty waly k sath bjh
```

```python
data["availability_rate"] = data["availability_365"] / 365
# ab avaliabilty rae k liye iskw 365 se dev krdety hyn
```
