```python
import nltk
from nltk.corpus import stopwords #Natrual Language Tool kit
nltk.download('stopwords')
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import re # Regural Expression


```

```python
def text_preprocessor(text):
  text = text.lower()
  text = "".join([a for a in text if a not in string.punctuation])
  text = " ".join([a for a in text.split() if a not in stopwords.words("english")])
  text = re.sub("\s+"," ",text)
  text = re.sub("\d+","",text)
  return text
```

```python
text_preprocessor("hell$o    sa089mi")
```

```python
data["reviews"]=data["reviews"].apply(text_preprocessor)
```
