```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def grade(self):
        if self.marks >= 85:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 50:
            return "C"
        else:
            return "Fail"

    def info(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Grade: {self.grade()}")


student1 = Student("Faheem", 86)
student1.info()
```

3. Create a class 'ElectricityBill' for a KE-electric consumer. Include units consumed and calculate
bill based on slab rates (e.g., different price per unit for different ranges).

```python
class ElectricityBill:
    def __init__(self,customer_name, units):
        self.customer_name = customer_name
        self.units = units

    def calculate_bill(self):
        if self.units <= 100:
            bill = self.units * 10
        elif self.units <= 200:
            bill = (100 * 10) + ((self.units - 100) * 15)
        else:
            bill = (100 * 10) + (100 * 15) + ((self.units - 200) * 20)


        return bill


    def display_bill(self):
        print(f"Consumer Name: {self.customer_name}")
        print(f"Units Consumed: {self.units}")
        print(f"Total Bill: Rs. {self.calculate_bill()}")

customer1 = ElectricityBill("Faheem" , 260)
customer1.display_bill()
```

6. Design an abstract class 'Shape' with a method 'area()'. Create subclasses 'Circle' and
'Rectangle' and implement area calculation logic.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle Area:", circle.area())
print("Rectangle Area:", rectangle.area())
```

7. Create a class 'HospitalPatient' with attributes name, age, and disease. Add a method to
determine priority (Emergency if age > 60 or serious disease).

```python
class HospitalPatient:
    def __init__(self, name, age, disease):
        self.name = name
        self.age = age
        self.disease = disease

    def priority(self):
        serious_diseases = ["heart attack", "stroke", "cancer", "kidney failure"]

        if self.age >= 60 or self.disease.lower() in serious_diseases:
            return "Emergency"
        else:
            return "Normal"

p1 = HospitalPatient("Ali", 65, "fever")
print(p1.priority())
```

8. Build a class 'ShoppingCart' where user can add items with price. Implement logic to calculate
total bill and apply discount if total exceeds a certain amount.

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_items(self,name,price):
        self.items.append((name,price))
    def total_bill(self):
        return sum(price for name , price in self.items)

    def final_bill(self):
        total = self.total_bill()

        discount = total * 0.10
        total -= discount
        return total

cart = ShoppingCart()
cart.add_items("shoe",10000)
cart.add_items("shirt",2000)

print("TotalBill", cart.total_bill())
print("FinalBill", cart.final_bill())
```

10. Design a class 'Library' where users can borrow and return books. Add logic to check
availability before issuing a book.

```python
class Library:
    def __init__(self):
        self.books = ["Python", "Java", "C++", "JavaScript"]

    def borrowed_books(self,book):
        if book in self.books:
            self.books.remove(book)
            print(f"you have borrowed '{book}'.")
        else:
            print(f"sorry its not available right now")

    def return_books(self, book):
        self.books.append(book)
        print(f"you have returned ''{book}'.")

    def show_books(self):
        print("Available books:", self.books)

library = Library()

library.show_books()

library.borrowed_books("Python")
library.show_books()

library.borrowed_books("Python")   

library.return_books("Python")
library.show_books()
```

```python
from abc import ABC, abstractmethod
import uuid


# Abstract Class
class Flight(ABC):
    def __init__(self, price, seats):
        self.__ticket_price = price      # Encapsulation
        self.seats = seats

    def get_price(self):
        return self.__ticket_price

    def set_price(self, price):
        self.__ticket_price = price

    @abstractmethod
    def flight_type(self):
        pass


# Subclasses
class Economy(Flight):
    def flight_type(self):
        return "Economy"


class Business(Flight):
    def flight_type(self):
        return "Business"


class FirstClass(Flight):
    def flight_type(self):
        return "First Class"


# Passenger Class
class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def book_ticket(self, flight, installments=False):
        if flight.seats <= 0:
            print("No seats available!")
            return

        # Reduce seat count
        flight.seats -= 1

        price = flight.get_price()

        # Discount
        if self.age < 12:
            price *= 0.50      # 50% discount for children
        elif self.age >= 60:
            price *= 0.80      # 20% discount for senior citizens

        # Airport tax (12%)
        tax = price * 0.12
        total = price + tax

        booking_id = str(uuid.uuid4())[:8]

        print("\n------ TICKET ------")
        print("Booking ID :", booking_id)
        print("Passenger  :", self.name)
        print("Age        :", self.age)
        print("Class      :", flight.flight_type())
        print("Base Price :", round(price, 2))
        print("Airport Tax:", round(tax, 2))
        print("Total Bill :", round(total, 2))

        if installments:
            print("Installment Plan: 3 Months")
            print("Monthly Payment :", round(total / 3, 2))
        else:
            print("Payment Mode: Full Payment")

        print("Seats Left :", flight.seats)
        print("---------------------")

        # Save booking to file
        with open("bookings.txt", "a") as file:
            file.write(
                f"{booking_id}, {self.name}, {self.age}, "
                f"{flight.flight_type()}, {round(total,2)}\n"
            )


# Example Usage
eco = Economy(10000, 2)

p1 = Passenger("Ali", 10)
p1.book_ticket(eco)

p2 = Passenger("Ahmed", 65)
p2.book_ticket(eco, installments=True)
```

```python
from abc import ABC, abstractmethod
import uuid


# Abstract Class
class Product(ABC):
    def __init__(self, name, price, stock):
        self.name = name
        self.__price = price      # Encapsulation
        self.stock = stock

    def get_price(self):
        return self.__price

    def set_price(self, price):
        self.__price = price

    @abstractmethod
    def category(self):
        pass


# Subclasses
class AC(Product):
    def category(self):
        return "Air Conditioner"


class Refrigerator(Product):
    def category(self):
        return "Refrigerator"


class WashingMachine(Product):
    def category(self):
        return "Washing Machine"


# Customer Class
class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, product):
        if product.stock > 0:
            self.cart.append(product)
            product.stock -= 1
            print(f"{product.name} added to cart.")
        else:
            print(f"{product.name} is out of stock!")

    def checkout(self, installment=False):
        subtotal = 0

        for product in self.cart:
            price = product.get_price()

            # Discount: 10% if price > 50000
            if price > 50000:
                price *= 0.90

            subtotal += price

        tax = subtotal * 0.18
        total = subtotal + tax

        invoice_id = str(uuid.uuid4())[:8]

        print("\n========== INVOICE ==========")
        print("Invoice ID:", invoice_id)
        print("Customer:", self.name)
        print("\nProducts:")

        for product in self.cart:
            print("-", product.name, f"({product.category()})")

        print("\nSubtotal:", round(subtotal, 2))
        print("Tax (18%):", round(tax, 2))
        print("Total:", round(total, 2))

        if installment:
            print("Payment: Installments")
            print("Monthly (3 months):", round(total / 3, 2))
        else:
            print("Payment: Full")

        print("=============================\n")

        # Save invoice
        with open("shop_records.txt", "a") as file:
            file.write(
                f"{invoice_id}, {self.name}, "
                f"Total: {round(total,2)}\n"
            )


# -------------------------
# Example Usage
# -------------------------

ac = AC("Haier AC", 80000, 5)
fridge = Refrigerator("Dawlance Fridge", 60000, 2)
wm = WashingMachine("LG Washing Machine", 40000, 1)

customer = Customer("Ali")

customer.add_to_cart(ac)
customer.add_to_cart(fridge)
customer.add_to_cart(wm)

customer.checkout(installment=True)
```

```python
from abc import ABC, abstractmethod
import uuid


# Abstract Class
class FoodItem(ABC):
    def __init__(self, name, price, quantity):
        self.name = name
        self.__price = price      # Encapsulation
        self.__quantity = quantity

    def get_price(self):
        return self.__price

    def get_quantity(self):
        return self.__quantity

    def set_quantity(self, quantity):
        self.__quantity = quantity

    @abstractmethod
    def category(self):
        pass


# Subclasses
class Pizza(FoodItem):
    def category(self):
        return "Pizza"


class Burger(FoodItem):
    def category(self):
        return "Burger"


class Drink(FoodItem):
    def category(self):
        return "Drink"


# Customer Class
class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, item):
        if item.get_quantity() > 0:
            self.cart.append(item)
            item.set_quantity(item.get_quantity() - 1)
            print(f"{item.name} added to cart.")
        else:
            print(f"{item.name} is unavailable!")

    def checkout(self, payment_method):
        subtotal = 0

        for item in self.cart:
            subtotal += item.get_price()

        # Discount
        discount = 0
        if subtotal > 2000:
            discount = subtotal * 0.10

        subtotal -= discount

        # Tax
        tax = subtotal * 0.15
        total = subtotal + tax

        order_id = str(uuid.uuid4())[:8]

        print("\n========== FOOD INVOICE ==========")
        print("Order ID:", order_id)
        print("Customer:", self.name)

        print("\nItems Ordered:")
        for item in self.cart:
            print("-", item.name, f"({item.category()})")

        print("\nDiscount:", round(discount, 2))
        print("Tax (15%):", round(tax, 2))
        print("Total Bill:", round(total, 2))
        print("Payment Method:", payment_method)
        print("==================================")

        # Save Order
        with open("food_orders.txt", "a") as file:
            file.write(
                f"{order_id}, {self.name}, "
                f"Payment: {payment_method}, "
                f"Total: {round(total,2)}\n"
            )


# -------------------------
# Example Usage
# -------------------------

pizza = Pizza("Chicken Fajita Pizza", 1500, 5)
burger = Burger("Zinger Burger", 700, 3)
drink = Drink("Pepsi", 200, 10)

customer = Customer("Ali")

customer.add_to_cart(pizza)
customer.add_to_cart(burger)
customer.add_to_cart(drink)

customer.checkout("Card")
```
